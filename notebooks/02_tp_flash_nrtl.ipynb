# TP flash: NRTL (gamma-phi)

Gamma-phi TP flash with NRTL for the liquid and Peng-Robinson for the vapor.


```python
import chemthermo as ct

temperature_K = 240.0
pressure_Pa = 3.0e6

names = ("Methane", "Ethane")
components = tuple(ct.Component.from_database(name) for name in names)
z = (0.50, 0.50)

mixture = ct.Mixture(
    components=components,
    composition=ct.Composition(fractions=z, basis="mole", normalize=False),
)

eos = ct.PengRobinsonEOS()
activity_model = ct.NRTL(parameters=ct.ActivityParameters.load("NRTL"))

result = ct.flash_tp(
    mixture,
    temperature_K=temperature_K,
    pressure_Pa=pressure_Pa,
    eos=eos,
    activity_model=activity_model,
    flash_mode="gamma-phi",
)

print("Phase names:", result.phase_names())
print("Vapor fraction beta:", result.vapor_fraction)

liquid = result.phases.get("liquid")
vapor = result.phases.get("vapor")
if liquid:
    print("x (liquid):", liquid.composition.fractions)
if vapor:
    print("y (vapor):", vapor.composition.fractions)

if liquid and vapor:
    ks = [y / x if x > 0.0 else float('inf') for x, y in zip(liquid.composition.fractions, vapor.composition.fractions)]
    print("K-values (y/x):", ks)
```
